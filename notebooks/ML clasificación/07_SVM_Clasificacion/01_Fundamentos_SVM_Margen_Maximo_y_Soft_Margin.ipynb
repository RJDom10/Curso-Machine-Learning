# 01 - Fundamentos de Support Vector Machines (SVM): Margen Máximo y Soft Margin

## Curso de Machine Learning: Modelos de Clasificación

Las Máquinas de Vectores de Soporte (SVM) son uno de los modelos más robustos en Machine Learning. Su objetivo fundamental es encontrar el **hiperplano óptimo** que maximice el margen de separación entre distintas clases.

---

### Contenidos
1. El concepto de Hiperplano Separador y Margen Máximo.
2. Vectores de Soporte: Las instancias que definen la frontera.
3. *Hard Margin* vs *Soft Margin* (Margen Suave).
4. El parámetro de regularización $C$ y las variables de holgura ($\xi$).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import make_blobs
from sklearn.inspection import DecisionBoundaryDisplay

plt.style.use('seaborn-v0_8-whitegrid')
np.random.seed(42)
```

## 1. El Hiperplano de Margen Máximo y Vectores de Soporte
Entrenaremos un modelo SVM con kernel lineal en un dataset perfectamente separable (Hard Margin) para visualizar la frontera y los márgenes.


```python
X, y = make_blobs(n_samples=100, centers=2, random_state=6, cluster_std=1.2)

# Entrenar SVM con Kernel Lineal y C alto (Hard Margin)
clf = SVC(kernel='linear', C=1000)
clf.fit(X, y)

plt.figure(figsize=(8, 5))
plt.scatter(X[:, 0], X[:, 1], c=y, s=30, cmap='coolwarm')

# Trazar la frontera de decision y los margenes
ax = plt.gca()
DecisionBoundaryDisplay.from_estimator(
    clf, X, plot_method='contour',
    colors='k', levels=[-1, 0, 1], alpha=0.5,
    linestyles=['--', '-', '--'], ax=ax
)

# Resaltar los vectores de soporte
ax.scatter(clf.support_vectors_[:, 0], clf.support_vectors_[:, 1], s=150,
           linewidth=1.5, facecolors='none', edgecolors='k', label='Vectores de Soporte')
plt.title('SVM Lineal: Hiperplano de Margen Maximo', fontweight='bold')
plt.legend()
plt.show()
```

## 2. El Parámetro $C$: Hard Margin vs Soft Margin
En la práctica, los datos no son perfectamente separables. SVM introduce variables de holgura ($\xi$) para permitir violaciones del margen, penalizadas por el parámetro **$C$**.

- **$C$ muy alto**: Penalización severa por errores. Margen más estrecho (Hard Margin). Riesgo de sobreajuste.
- **$C$ muy bajo**: Penalización leve. Margen más amplio (Soft Margin), permite más violaciones. Mayor regularización.


```python
# Generamos datos con solapamiento
X_soft, y_soft = make_blobs(n_samples=150, centers=2, random_state=4, cluster_std=1.5)

C_values = [0.01, 100]
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, C in zip(axes, C_values):
    clf_soft = SVC(kernel='linear', C=C).fit(X_soft, y_soft)
    
    ax.scatter(X_soft[:, 0], X_soft[:, 1], c=y_soft, s=30, cmap='coolwarm')
    DecisionBoundaryDisplay.from_estimator(
        clf_soft, X_soft, plot_method='contour',
        colors='k', levels=[-1, 0, 1], alpha=0.5,
        linestyles=['--', '-', '--'], ax=ax
    )
    ax.scatter(clf_soft.support_vectors_[:, 0], clf_soft.support_vectors_[:, 1], s=100,
               linewidth=1, facecolors='none', edgecolors='k')
    ax.set_title(f'Soft Margin SVM (C = {C})', fontweight='bold')

plt.show()
```
