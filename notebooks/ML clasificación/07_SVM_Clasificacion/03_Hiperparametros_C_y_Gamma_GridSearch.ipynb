# 03 - Sintonización de Hiperparámetros: C y Gamma

## Curso de Machine Learning: Modelos de Clasificación

En un SVM con Kernel RBF, los dos hiperparámetros más críticos que controlan el dilema Sesgo-Varianza son:
- **$C$ (Regularización)**: Penalización por clasificaciones erróneas. $C$ alto = frontera estricta (baja varianza teórica del error en train, alto riesgo de sobreajuste). $C$ bajo = frontera suave.
- **$\gamma$ (Gamma)**: Define la 'influencia' o el radio de alcance de un solo vector de soporte. $\gamma$ alto = influencia muy local (fronteras complejas/fragmentadas). $\gamma$ bajo = influencia global (fronteras suaves/casi lineales).


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import load_breast_cancer
from sklearn.model_selection import train_test_split, GridSearchCV
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline

plt.style.use('seaborn-v0_8-whitegrid')
```

## Búsqueda Exhaustiva con GridSearchCV
Las SVM son extremadamente sensibles a la escala de los datos, por lo que **SIEMPRE** deben escalarse (`StandardScaler`) antes de entrenar.


```python
cancer = load_breast_cancer()
X_train, X_test, y_train, y_test = train_test_split(cancer.data, cancer.target, test_size=0.25, random_state=42)

pipe = Pipeline([
    ('scaler', StandardScaler()),
    ('svc', SVC(kernel='rbf', random_state=42))
])

param_grid = {
    'svc__C': [0.1, 1, 10, 100],
    'svc__gamma': [1, 0.1, 0.01, 0.001]
}

grid = GridSearchCV(pipe, param_grid, cv=5, n_jobs=-1, scoring='accuracy')
grid.fit(X_train, y_train)

print(f"Mejores Hiperparametros: {grid.best_params_}")
print(f"Exactitud en CV: {grid.best_score_*100:.2f}%")
print(f"Exactitud en Test: {grid.score(X_test, y_test)*100:.2f}%")
```
