# 02 - El Truco del Kernel (Kernel Trick) y Transformaciones No Lineales

## Curso de Machine Learning: Modelos de Clasificación

Cuando los datos no son separables por una línea recta (o un hiperplano plano en $N$-dimensiones), SVM utiliza el **Truco del Kernel**.
Esto permite mapear los datos a un espacio de mayor dimensionalidad donde sí sean linealmente separables, sin necesidad de computar explícitamente las coordenadas en ese nuevo espacio, ahorrando un inmenso coste computacional.


```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.datasets import make_circles
from sklearn.inspection import DecisionBoundaryDisplay

plt.style.use('seaborn-v0_8-whitegrid')
```

## 1. El Problema de la Separabilidad No Lineal


```python
X_circ, y_circ = make_circles(n_samples=300, factor=0.3, noise=0.1, random_state=42)

# Intento fallido con Kernel Lineal
clf_lin = SVC(kernel='linear').fit(X_circ, y_circ)

plt.figure(figsize=(6, 5))
plt.scatter(X_circ[:, 0], X_circ[:, 1], c=y_circ, cmap='coolwarm', s=30)
DecisionBoundaryDisplay.from_estimator(clf_lin, X_circ, plot_method='contour', colors='k', levels=[0], ax=plt.gca())
plt.title('SVM Lineal fracasa en datos circulares', fontweight='bold')
plt.show()
```

## 2. Aplicando Kernels: RBF y Polinomial
- **RBF (Radial Basis Function)**: También conocido como Kernel Gaussiano. Es el estándar por defecto. Mapea los datos a un espacio de dimensión infinita.
- **Polynomial**: Mapea a un espacio polinómico de grado $d$.


```python
kernels = ['poly', 'rbf']
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

for ax, k in zip(axes, kernels):
    # degree=3 solo aplica para 'poly'
    clf_k = SVC(kernel=k, C=1.0, degree=3, gamma='scale').fit(X_circ, y_circ)
    
    DecisionBoundaryDisplay.from_estimator(
        clf_k, X_circ, cmap='coolwarm', alpha=0.6, ax=ax, response_method='predict'
    )
    ax.scatter(X_circ[:, 0], X_circ[:, 1], c=y_circ, cmap='coolwarm', edgecolors='k', s=30)
    ax.set_title(f'SVM con Kernel: {k.upper()}', fontweight='bold')

plt.show()
```
