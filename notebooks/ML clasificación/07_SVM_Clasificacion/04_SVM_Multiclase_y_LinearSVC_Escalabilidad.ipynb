# 04 - SVM Multiclase y Escalabilidad (LinearSVC)

## Curso de Machine Learning: Modelos de Clasificación

Por naturaleza, el algoritmo SVM es estrictamente **binario**. Para clasificar múltiples clases ($K > 2$), Scikit-Learn utiliza internamente estrategias probabilísticas:
- **OvO (One-vs-One)**: Entrena $K(K-1)/2$ clasificadores binarios. Es el método por defecto de `SVC(kernel='rbf')`.
- **OvR (One-vs-Rest / One-vs-All)**: Entrena $K$ clasificadores. Es el método estándar de `LinearSVC`.


```python
import time
from sklearn.datasets import make_classification
from sklearn.svm import SVC, LinearSVC
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

## Escalabilidad Computacional: `SVC` vs `LinearSVC`
La implementación matemática estándar de `SVC(kernel='linear')` está basada en *libsvm* y tiene una complejidad temporal de $O(N^2)$ a $O(N^3)$. Para datasets masivos (+$100,000$ muestras), es impracticable.
Scikit-Learn provee `LinearSVC` basado en *liblinear*, el cual escala de forma casi lineal $O(N)$.


```python
# Generar un dataset masivo (100,000 muestras)
X_large, y_large = make_classification(n_samples=50000, n_features=20, n_classes=2, random_state=42)
X_tr, X_te, y_tr, y_te = train_test_split(X_large, y_large, test_size=0.2)

# 1. Usando LinearSVC (Optimizado para linealidad y volumen)
t0 = time.time()
clf_lin = LinearSVC(dual=False, random_state=42).fit(X_tr, y_tr)
t_lin = time.time() - t0

# 2. Usando SVC estándar con kernel='linear'
t0 = time.time()
clf_svc = SVC(kernel='linear', random_state=42).fit(X_tr, y_tr)
t_svc = time.time() - t0

print("Comparativa de Tiempos de Entrenamiento:")
print(f"LinearSVC:             {t_lin:.3f} segundos (Acc: {accuracy_score(y_te, clf_lin.predict(X_te))*100:.1f}%)")
print(f"SVC(kernel='linear'):  {t_svc:.3f} segundos (Acc: {accuracy_score(y_te, clf_svc.predict(X_te))*100:.1f}%)")
```
