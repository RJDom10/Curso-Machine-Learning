# 02 - Algoritmo Random Forest en Clasificación

## Curso de Machine Learning: Modelos de Clasificación

El algoritmo **Random Forest (Bosques Aleatorios)**, propuesto por Leo Breiman en 2001, representa una de las mayores innovaciones en la historia del aprendizaje automático. Este algoritmo no es solo una extensión de Bagging, sino una modificación profunda diseñada para destruir matemáticamente la correlación entre los estimadores base.

En este cuaderno profundizaremos en la mecánica matemática de los nodos, la inyección de la doble aleatoriedad y la trampa analítica de la interpretabilidad basada en impureza.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.datasets import load_breast_cancer, make_moons
from sklearn.model_selection import train_test_split
from sklearn.ensemble import RandomForestClassifier, BaggingClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.inspection import DecisionBoundaryDisplay, permutation_importance
from sklearn.metrics import accuracy_score

plt.style.use('seaborn-v0_8-whitegrid' if 'seaborn-v0_8-whitegrid' in plt.style.available else 'default')
np.random.seed(42)
print("Entorno preparado para Random Forest.")
```

## 1. El Límite de Bagging y las Matemáticas de la Impureza

Como vimos en el cuaderno anterior, la varianza de un ensamble de árboles es: $\mathbb{V}ar = \rho \sigma^2 + \frac{1 - \rho}{B} \sigma^2$.
Bagging reduce el segundo término aumentando $B$ (el número de árboles), pero deja casi intacto el primer término $\rho \sigma^2$ (la covarianza o correlación entre árboles).

### ¿Por qué los árboles de Bagging se correlacionan?
En cada nodo, un árbol de decisión evalúa **todas las variables disponibles** y elige la que maximiza la ganancia de información (reducción de impureza). La impureza de Gini de un nodo se define como:

$$ I_G(p) = 1 - \sum_{i=1}^C p_i^2 $$

Donde $p_i$ es la proporción de la clase $i$ en ese nodo. Si existe una variable $X_j$ que es extremadamente buena separando las clases (es decir, reduce drásticamente $I_G$), **casi todos los árboles generados por Bagging elegirán $X_j$ en su nodo raíz**, sin importar las variaciones inducidas por la muestra Bootstrap.

El resultado es un "bosque" donde todos los árboles tienen estructuras casi idénticas en sus ramas principales. $\rho$ se mantiene alto, y la varianza no se reduce adecuadamente.


## 2. Random Forest: La Regla de la Doble Aleatoriedad

Leo Breiman resolvió este problema inyectando un **segundo nivel de aleatoriedad**: el muestreo de características (*Feature Subsampling* o *Feature Bagging*).

En lugar de buscar la mejor división entre las $D$ características totales del dataset, Random Forest obliga a cada árbol a buscar la mejor división dentro de un subconjunto aleatorio de tamaño $m$ en **cada uno de sus nodos**.

### La heurística de oro: $m = \sqrt{D}$
Para problemas de clasificación, la regla empírica universalmente aceptada (y el valor por defecto en `scikit-learn`) es:

$$ m = \lfloor \sqrt{D} \rfloor $$

**Impacto Matemático:**
- Si la variable "dominante" $X_j$ existe, solo estará disponible en una fracción de las divisiones.
- Esto **obliga** a los árboles a explorar variables secundarias o interacciones ocultas.
- Los árboles se vuelven estructuralmente diversos, disminuyendo drásticamente la correlación $\rho$, lo que colapsa el término $\rho \sigma^2$ de la varianza total.


```python
# Demostración visual de la descorrelación y el suavizado de la frontera de decisión
X_moons, y_moons = make_moons(n_samples=300, noise=0.35, random_state=42)

# Modelos a comparar
modelos = {
    'Árbol Individual': DecisionTreeClassifier(random_state=42),
    'Bagging (m=D, todos los features)': BaggingClassifier(estimator=DecisionTreeClassifier(), n_estimators=100, random_state=42),
    'Random Forest (m=sqrt(D))': RandomForestClassifier(n_estimators=100, max_features='sqrt', random_state=42)
}

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for ax, (nombre, modelo) in zip(axes, modelos.items()):
    modelo.fit(X_moons, y_moons)
    acc = modelo.score(X_moons, y_moons)
    
    DecisionBoundaryDisplay.from_estimator(
        modelo, X_moons, cmap='coolwarm', alpha=0.6, ax=ax, response_method='predict'
    )
    ax.scatter(X_moons[:, 0], X_moons[:, 1], c=y_moons, cmap='coolwarm', edgecolor='k', s=35)
    ax.set_title(f'{nombre}\nAccuracy en Entrenamiento: {acc*100:.1f}%', fontweight='bold')

plt.suptitle('Evolución de las Fronteras de Decisión: De la Rigidez Ortogonal a la Suavidad Probabilística', fontsize=14, fontweight='bold', y=1.05)
plt.tight_layout()
plt.show()

print("Análisis: El Árbol individual memoriza ruido (sobreajuste severo). Bagging suaviza un poco, pero mantiene geometrías cuadradas. Random Forest logra la frontera más orgánica.")
```

## 3. Comportamiento Asintótico: ¿Por qué Random Forest NO sobreajusta al crecer?

Un temor común en Machine Learning es que añadir complejidad (más capas en una red neuronal, o más iteraciones en Boosting) lleva irremediablemente al sobreajuste.

**Random Forest es una excepción fundamental por la Ley de los Grandes Números.**
A medida que añadimos más árboles (incrementamos `n_estimators`), el modelo no se vuelve más complejo en términos de capacidad de memorización; simplemente **suaviza su estimación estadística**. El error de generalización de un Bosque Aleatorio converge a un límite estricto.

Demostrémoslo graficando la curva de error Out-Of-Bag (OOB) vs Test.


```python
cancer = load_breast_cancer()
X_bc, y_bc = cancer.data, cancer.target
X_train, X_test, y_train, y_test = train_test_split(X_bc, y_bc, test_size=0.25, random_state=42, stratify=y_bc)

n_estimators_list = [10, 25, 50, 75, 100, 150, 200, 300, 400, 500]
oob_errors = []
test_errors = []

for n in n_estimators_list:
    rf = RandomForestClassifier(n_estimators=n, max_features='sqrt', oob_score=True, random_state=42, n_jobs=-1)
    rf.fit(X_train, y_train)
    oob_errors.append(1.0 - rf.oob_score_) # Error OOB
    test_errors.append(1.0 - rf.score(X_test, y_test)) # Error Test

plt.figure(figsize=(10, 5))
plt.plot(n_estimators_list, oob_errors, 'o-', color='#2b5c8f', lw=2.5, label='Error de Generalización Estimado (OOB)')
plt.plot(n_estimators_list, test_errors, 's-', color='#d95f02', lw=2.5, label='Error Real en Conjunto de Prueba (Test)')
plt.axvline(150, color='gray', linestyle='--', label='Punto de Saturación / Retornos Decrecientes')
plt.xlabel('Número de Árboles (n_estimators)', fontweight='bold')
plt.ylabel('Tasa de Error', fontweight='bold')
plt.title('Demostración Empírica: Random Forest es Inmune al Sobreajuste por n_estimators', fontweight='bold', fontsize=12)
plt.legend()
plt.tight_layout()
plt.show()
```

## 4. El Peligro Analítico: Sesgo en la Importancia de Características

Un error gravísimo y muy común en la industria es confiar ciegamente en `rf.feature_importances_` (conocido como **MDI - Mean Decrease Impurity** o Importancia Gini).

**El Problema Matemático:**
La métrica MDI calcula cuánta impureza redujo una variable cada vez que fue elegida para un nodo. Sin embargo, variables continuas o categóricas de alta cardinalidad (muchos valores únicos) ofrecen *matemáticamente* muchísimas más opciones de corte. El árbol, al intentar sobreajustar en su entrenamiento profundo, elegirá estas variables para "pulverizar" los datos, otorgándoles una importancia inflada y engañosa.

**La Solución Profesional (Permutation Importance):**
Evaluamos la importancia **post-entrenamiento**, en un conjunto de datos separado (Test). Se toma una columna, se "barajan" (permutan) sus valores aleatoriamente, destruyendo su relación con el *Target*. Si el rendimiento del modelo cae en picada, esa variable era crucial. Si el rendimiento no cambia, la variable era inútil.

Hagamos una demostración inyectando una variable de **puro ruido aleatorio continuo** al dataset.


```python
# Añadimos intencionalmente una columna de RUIDO CONTINUO ALEATORIO al dataset
np.random.seed(42)
ruido_train = np.random.normal(0, 1, X_train.shape[0]).reshape(-1, 1)
ruido_test = np.random.normal(0, 1, X_test.shape[0]).reshape(-1, 1)

X_train_ruido = np.hstack([X_train, ruido_train])
X_test_ruido = np.hstack([X_test, ruido_test])
nombres_features = list(cancer.feature_names) + ['RUIDO_ALEATORIO (No Informa)']

# Entrenamos el Random Forest
rf_bias = RandomForestClassifier(n_estimators=100, max_features='sqrt', random_state=42, n_jobs=-1)
rf_bias.fit(X_train_ruido, y_train)

# 1. Importancia Nativa (MDI - Gini)
mdi_importances = pd.Series(rf_bias.feature_importances_, index=nombres_features)

# 2. Importancia Robusta por Permutación (en Test)
perm_results = permutation_importance(rf_bias, X_test_ruido, y_test, n_repeats=10, random_state=42, n_jobs=-1)
perm_importances = pd.Series(perm_results.importances_mean, index=nombres_features)

# Gráficas Comparativas
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

mdi_importances.sort_values(ascending=True).tail(12).plot(kind='barh', ax=axes[0], color='#2b5c8f')
axes[0].set_title('1. MDI (Impureza Gini - Entrenado)\n¡Falso Positivo! El Ruido parece importante.', fontweight='bold')
axes[0].set_xlabel('Reducción media de impureza')

perm_importances.sort_values(ascending=True).tail(12).plot(kind='barh', ax=axes[1], color='#d95f02')
axes[1].set_title('2. Importancia por Permutación (Evaluado en Test)\nEl Ruido es detectado como inútil o dañino.', fontweight='bold')
axes[1].set_xlabel('Caída media en Accuracy al permutar')

plt.tight_layout()
plt.show()
```

## 5. Conclusiones Directivas sobre Random Forest

1. **Doble Aleatoriedad:** Random Forest triunfa porque ataca ambos frentes estadísticos. El Bootstrap diversifica las muestras (reduciendo la varianza inicial), y la restricción de características ($m = \sqrt{D}$) destruye la correlación estructural de los árboles.
2. **Punto de Saturación:** Añadir árboles (`n_estimators`) nunca causará sobreajuste. Sin embargo, más allá de 150-200 árboles, los retornos marginales son prácticamente nulos, y solo incrementamos el costo computacional.
3. **Evaluación Interna:** El error `oob_score_` es tu mejor amigo para iterar rápido sin depender de un set de validación cruzada.
4. **Rigor en la Interpretabilidad:** En entornos de negocio altamente auditables (salud, finanzas), **nunca** reportes directamente `feature_importances_` (MDI) si tu dataset mezcla variables continuas, ordinales y categóricas. Usa siempre **Permutation Importance** sobre un conjunto de pruebas (*Holdout*) para extraer el verdadero valor predictivo.
