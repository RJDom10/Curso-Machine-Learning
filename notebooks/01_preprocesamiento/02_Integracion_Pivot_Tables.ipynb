## 1. Introduccion a la Integracion de Datos

La integracion de datos comprende la fusion de informacion proveniente de multiples fuentes o la reestructuracion de una misma fuente para consolidar la informacion en una forma que sea apta para el analisis o el modelado de Machine Learning.

Este proceso debe realizarse con estricto rigor analitico para evitar redundancias, inconsistencias y perdida de informacion en el conjunto de datos resultante.

Las operaciones tipicas logradas dentro de la integracion de datos son:
* **Identificacion y unificacion** de variables y dominios.
* **Analisis de correlacion** entre atributos.
* **Identificacion de duplicacion** y **deteccion de conflictos** en los valores de datos de diferentes representaciones.
* **Agregacion y reestructuracion** de datos a diferentes niveles de granularidad (por ejemplo, pasar de datos diarios a datos mensuales).

Una de las herramientas mas poderosas en el ecosistema de Pandas para la reestructuracion y agregacion de datos es la **Tabla Pivote** (`pivot_table`).

## 2. Fundamentos de las Tablas Pivote

Una tabla pivote permite resumir y reorganizar datos dinamicamente. Su objetivo es transformar filas y columnas para agrupar los datos y aplicar funciones de agregacion (suma, promedio, maximo, etc.) sobre subconjuntos de interes.

En el contexto de la preparacion de datos, las tablas pivote son invaluables para:
1. **Cambiar la granularidad del dataset:** Por ejemplo, agrupar transacciones individuales en ventas totales por mes y por cliente.
2. **Generar nuevas variables (Feature Engineering):** Extraer metricas estadisticas que capturen el comportamiento historico de un sujeto de estudio.
3. **Manejar series de tiempo con multiples sujetos:** Estructurar datos longitudinales de forma que cada sujeto sea una columna y cada instante de tiempo sea una fila (o viceversa).

### Sintaxis Basica en Pandas

La funcion `pd.pivot_table()` toma como argumentos principales:
* `data`: El DataFrame original.
* `values`: La columna o columnas a agregar.
* `index`: Las columnas cuyos valores se convertiran en el indice (filas) de la nueva tabla.
* `columns`: Las columnas cuyos valores se convertiran en las columnas de la nueva tabla.
* `aggfunc`: La funcion de agregacion a aplicar (por defecto es `numpy.mean`).

## 3. Analisis del Caso: Propagacion de COVID-19

Dada la pandemia desencadenada por el COVID-19, el estudio epidemiologico de la propagacion del virus a nivel global requiere un manejo extensivo de series temporales por pais. 

Para ilustrar la utilidad de las tablas pivote en el analisis de este tipo de fenomenos, utilizaremos una base de datos historica de casos a nivel mundial.

Primero, procedemos a importar las librerias necesarias y cargar los datos desde nuestro directorio de almacenamiento local.

```python
# Importacion de librerias base
import pandas as pd
import numpy as np

# Configuracion de opciones de visualizacion de pandas
pd.set_option('display.max_columns', 20)
pd.set_option('display.width', 1000)

# Carga del dataset desde el repositorio
url = "https://raw.githubusercontent.com/RJDom10/Curso-Machine-Learning/refs/heads/main/data/covid19WW_2021.csv"
df_covid = pd.read_csv(url)
# Exploracion inicial de la estructura tabular original
df_covid.head()
```

### 3.1. Ejemplo Base: Agrupacion Basica

Para nuestro analisis exploratorio, supongamos que requerimos enfocarnos en los **casos confirmados promedio por pais y fecha**.

La base de datos actual puede presentar una granularidad a nivel de provincia/estado para ciertos paises. El uso de `pivot_table` nos permite consolidar y promediar estas observaciones.

*Nota:* Dependiendo de la estructura especifica de los datos y el objetivo epidemiologico, la agregacion mediante suma (`np.sum`) podria ser matematicamente mas apropiada para obtener los casos totales del pais, en lugar del promedio de sus estados. Para fines de este ejemplo demostrativo, replicaremos el uso de la media (`np.mean`).

```python
# Creacion de tabla pivote para promediar casos confirmados
df_ejemplo = pd.pivot_table(
    data=df_covid,
    index=['Date', 'Country'],
    values='Confirmed',
    aggfunc=np.mean
)

print("Datos agregados mediante tabla pivote (Primeros 15 registros):")
df_ejemplo.head(15)
```
